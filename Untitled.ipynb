```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
df = sns.load_dataset('titanic')

# Display beginning and end
display(df.head())
display(df.tail())

# Record shape and info
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
df.info()
```

```python
# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object', 'str', 'category', 'bool']).columns

print("Numerical Summary:")
display(df[num_cols].describe())
```

```python
print("Missing Values Count:")
missing_counts = df.isnull().sum()
print(missing_counts[missing_counts > 0])

print("\nMissing Values Percentage:")
missing_percent = (df.isnull().mean() * 100).round(2)
print(missing_percent[missing_percent > 0])
```

```python
# Make a copy before modifying
df_clean = df.copy()

# Drop columns with too many missing values or redundant information
cols_to_drop = ['deck', 'alive', 'who', 'adult_male', 'class', 'embark_town']
df_clean = df_clean.drop(columns=cols_to_drop)

# Impute 'age' with median (better for skewed data)
age_median = df_clean['age'].median()
df_clean['age'] = df_clean['age'].fillna(age_median)

# Impute 'embarked' with mode (most frequent port)
embarked_mode = df_clean['embarked'].mode()[0]
df_clean['embarked'] = df_clean['embarked'].fillna(embarked_mode)

# Verify missing values are resolved
print("Remaining missing values:", df_clean.isnull().sum().sum()) 
```

```python
# Identify duplicates
duplicates = df_clean.duplicated().sum()
print(f"Duplicate rows found: {duplicates}")

# Remove duplicates
df_clean = df_clean.drop_duplicates()

# Check categorical consistency
print("Sex categories:", df_clean['sex'].unique())
print("Embarked categories:", df_clean['embarked'].unique())
```

```python
plt.figure(figsize=(12, 5))

# Visualisation 1: Histogram of Age
plt.subplot(1, 2, 1)
sns.histplot(df_clean['age'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')

# Visualisation 2: Bar chart of Survival by Sex
plt.subplot(1, 2, 2)
sns.countplot(data=df_clean, x='sex', hue='survived', palette='pastel')
plt.title('Survival Count by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Apply One-Hot Encoding to categorical variables
df_encoded = pd.get_dummies(df_clean, columns=['sex', 'embarked'], drop_first=True)

# Convert boolean columns to integers (0 and 1) for machine learning
bool_cols = df_encoded.select_dtypes(include=['bool']).columns
df_encoded[bool_cols] = df_encoded[bool_cols].astype(int)

display(df_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
cols_to_scale = ['age', 'fare']

# Apply standard scaling (mean=0, variance=1)
df_encoded[cols_to_scale] = scaler.fit_transform(df_encoded[cols_to_scale])

display(df_encoded.head())
```

```python
# Verify final shape and nulls
print(f"Final shape: {df_encoded.shape}")
print(f"Remaining nulls: {df_encoded.isnull().sum().sum()}")

# Save to CSV
final_filename = 'titanic_prepared.csv'
df_encoded.to_csv(final_filename, index=False)
print(f"Dataset successfully saved as '{final_filename}'")
```
